```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(options=options)

positions = ["Data Engineer", "Ai Engineer", "Data Analyst", "Data Scientist"]
list_jobs = []

for position in positions:
    print(f"\n=================== Starting Search For: {position} ===================")
    
    driver.get('https://www.naukrigulf.com/?showQsb=1')
    time.sleep(3)
    
    search_input = driver.find_element(By.ID, "qsbKey")
    search_input.clear()
    search_input.send_keys(position)
    
    time.sleep(1)
    driver.find_element(By.CSS_SELECTOR, "input[value='Search Jobs']").click()
    time.sleep(3)
    
    for page in range(3):
        print(f"[{position}] Scraped Page {page + 1}...")
        time.sleep(4)  
        
        jobs = driver.find_elements(By.CLASS_NAME, "ng-box")
        
        for job in jobs:
            try:
                job_title = job.find_element(By.CLASS_NAME, "designation-title").text
            except:
                job_title = None
                
            try:
                company_element = job.find_element(By.CLASS_NAME, "info-org")
                company = company_element.get_attribute("title")
                if not company:
                    company = company_element.text
            except:
                company = None
                
            try:
                location = job.find_element(By.CLASS_NAME, "info-loc").text
            except:
                location = None
               
            try:
                experience = job.find_element(By.CLASS_NAME, "info-exp").text
            except:
                experience = None
                
            try:
                description = job.find_element(By.CLASS_NAME, "description").text
            except:
                description = None
            
          
            list_jobs.append({             
                "searched_keyword": position,
                "job_title": job_title,
                "company": company, 
                "location": location,
                "experience": experience,
                "description": description            
            })
        
     
        if page < 2:
            try:
                next_button = WebDriverWait(driver, 5).until(
                    EC.element_to_be_clickable((By.CSS_SELECTOR, ".ico.bwd"))
                )
                driver.execute_script("arguments[0].click();", next_button)
            except Exception as e:
                print(f"انتهت الصفحات المتاحة لـ {position} أو حدث خطأ: {e}")
                break

driver.quit()

print(f"\nتم تجميع إجمالي {len(list_jobs)} وظيفة لجميع المسميات بنجاح!")
```

```python
import pandas as pd
import re

# قائمة بالمهارات الأكثر طلباً في مجالات البيانات والذكاء الاصطناعي
SKILLS_LIST = [
    'Python', 'SQL', 'Power BI', 'Tableau', 'Excel', 'R', 'ETL', 'Spark', 'PySpark',
    'AWS', 'Azure', 'GCP', 'Docker', 'Kubernetes', 'Airflow', 'Snowflake', 'Databricks',
    'Machine Learning', 'Deep Learning', 'Pandas', 'NumPy', 'Scikit-learn', 'TensorFlow',
    'PyTorch', 'Git', 'Kafka', 'Data Warehouse', 'DAX', 'Power Query'
]

def extract_skills(description):
    if not description:
        return "Not Specified"
    found_skills = []
    for skill in SKILLS_LIST:
        # البحث عن الكلمة مع مراعاة حالة الأحرف وتفادي الكلمات الجزئية
        if re.search(r'\b' + re.escape(skill) + r'\b', description, re.IGNORECASE):
            found_skills.append(skill)
    return ", ".join(found_skills) if found_skills else "Other"

# تحويل البيانات إلى DataFrame واستخراج المهارات
df = pd.DataFrame(list_jobs)
df.drop_duplicates(inplace=True)
df['extracted_skills'] = df['description'].apply(extract_skills)

# تصدير الملف لـ Power BI
df.to_csv("naukrigulf_jobs_with_skills.csv", index=False, encoding="utf-8-sig")
print("تم استخراج المهارات بنجاح وحفظ البيانات!")
```
